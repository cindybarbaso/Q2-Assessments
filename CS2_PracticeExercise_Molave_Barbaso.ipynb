```python
# Exercise 1 - Personal savings tracker

def update_savings(current_total, deposit):
    return current_total + deposit

def check_goal(total, target):
    if total >= target:
        return "Goal reached!"
    return "Keep saving!"

target = float(input("Enter savings target: ₱"))
total = 0.0

while True:
    deposit = float(input("Enter deposit (0 to finish): ₱"))
    if deposit == 0:
        break
    total = update_savings(total, deposit)
    print(f"Updated savings: ₱{total:.2f}")

print(f"Total savings: ₱{total:.2f}")
print(check_goal(total, target))
```

```python
# Exercise 2 - Classroom energy usage analyzer
def calculate_energy(power_watts, hours_used):
    return (power_watts * hours_used) / 1000

def classify_usage(energy):
    if energy < 0.5:
        return "Low"
    if energy <= 2.0:
        return "Moderate"
    return "High"

devices = int(input("\nEnter number of devices to analyze: "))
total_kwh = 0.0
low = 0
mod = 0
high = 0

for i in range(1, devices + 1):
    print(f"\nDevice {i}")
    name = input("Enter device name: ")
    w = float(input("Enter power (Watts): "))
    h = float(input("Enter hours used: "))

    kwh = calculate_energy(w, h)
    status = classify_usage(kwh)
    total_kwh += kwh

    if status == "Low":
        low += 1
    elif status == "Moderate":
        mod += 1
    else:
        high += 1

    print(f"Device: {name}, Consumption: {kwh:.2f} kWh, Class: {status}")

print("Classroom energy summary")
print(f"Total daily energy Consumed: {total_kwh:.2f} kWh")
print(f"Low impact Devices: {low}")
print(f"Moderate impact Devices: {mod}")
print(f"High impact Devices: {high}")
```

```python
# Excercise 3 - Water refilling station sales tracker

def calculate_sale(quantity, price_per_container):
    return quantity * price_per_container

def calculate_discount(total):
    if total >= 500:
        return total * 0.05
    return 0.0

price = float(input("Price per container: ₱"))
total_gross = 0.0
total_discounts = 0.0

while True:
    qty = int(input("Enter quantity sold (0 to finish): "))
    if qty == 0:
        break

    sale = calculate_sale(qty, price)
    discount = calculate_discount(sale)
    due = sale - discount

    total_gross += sale
    total_discounts += discount

    print(f"Sale: ₱{sale:.2f}")
    print(f"Discount: ₱{discount:.2f}")
    print(f"Amount due: ₱{due:.2f}")

print(f"Total sales before discounts: ₱{total_gross:.2f}")
print(f"Total discounts: ₱{total_discounts:.2f}")
print(f"Net sales: ₱{total_gross - total_discounts:.2f}")
```

```python
# Exercise 4 - Reading time and progress analyzer

def calculate_reading_speed(pages, minutes):
    if minutes == 0:
        return 0
    return pages / minutes

def classify_speed(speed):
    if speed < 1.0:
        return "Developing"
    if speed <= 2.0:
        return "Steady"
    return "Fast"

sessions = int(input("\nHow many reading sessions? "))
total_pages = 0
total_minutes = 0

for i in range(1, sessions + 1):
    pages = int(input(f"\nSession {i} pages read: "))
    minutes = int(input("Minutes spent: "))

    speed = calculate_reading_speed(pages, minutes)
    status = classify_speed(speed)

    total_pages += pages
    total_minutes += minutes

    print(f"Reading speed: {speed:.2f} pages/minute")
    print(f"Classification: {status}")

overall_speed = calculate_reading_speed(total_pages, total_minutes)
print(f"\nTotal pages read: {total_pages}")
print(f"Total reading time: {total_minutes} minutes")
print(f"Overall reading speed: {overall_speed:.2f} pages/minute")
```

```python
# Exercise 5 - School event attendance monitor

def calculate_attendance_rate(attendees, expected):
    if expected == 0:
        return 0
    return (attendees / expected) * 100

def classify_attendance(rate):
    if rate < 60:
        return "Low"
    if rate < 90:
        return "Moderate"
    return "High"

sessions_count = int(input("\nHow many sessions? "))
total_attendees = 0
total_expected = 0
high_sessions = 0

for i in range(1, sessions_count + 1):
    expected = int(input(f"Session {i} expected attendance: "))
    actual = int(input("Actual attendees: "))

    rate = calculate_attendance_rate(actual, expected)
    status = classify_attendance(rate)

    total_attendees += actual
    total_expected += expected
    if status == "High":
        high_sessions += 1

    print(f"Attendance rate: {rate:.2f}%")
    print(f"Classification: {status}")

overall_rate = calculate_attendance_rate(total_attendees, total_expected)
print(f"Total attendees: {total_attendees}")
print(f"Total expected attendance: {total_expected}")
print(f"Overall attendance rate: {overall_rate:.2f}%")
print(f"High-attendance sessions: {high_sessions}")
```
