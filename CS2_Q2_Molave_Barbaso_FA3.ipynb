```python
# Student Score Analyzer

students = ["John", "Mary", "Peter", "Anna", "Mark"]
scores = [85, 90, 78, 95, 88]


def display_scores():
    print("\nStudent Scores:")
    for i in range(len(students)):
        print(students[i], ":", scores[i])


def search_score():
    name = input("Enter student name: ")

    if name in students:
        index = students.index(name)
        print("Name:", students[index])
        print("Score:", scores[index])
    else:
        print("Student not found!")


def compute_average():
    average = sum(scores) / len(scores)
    print("Average Score:", average)


def update_score():
    name = input("Enter student name to update: ")

    if name in students:
        index = students.index(name)
        print("Current Score:", scores[index])

        new_score = int(input("Enter new score: "))
        scores[index] = new_score

        print("Score updated successfully!")
    else:
        print("Student not found!")


def display_updated_scores():
    print("\nUpdated Student Scores:")
    for i in range(len(students)):
        print(students[i], ":", scores[i])


def main():
    while True:
        print("\nStudent Score Analyzer")
        print("Select the number of the operation you want to perform:")
        print("1. Display Scores")
        print("2. Search Score")
        print("3. Compute Average Score")
        print("4. Update Score")
        print("5. Display Updated Scores")
        print("6. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            display_scores()
        elif choice == "2":
            search_score()
        elif choice == "3":
            compute_average()
        elif choice == "4":
            update_score()
        elif choice == "5":
            display_updated_scores()
        elif choice == "6":
            print("Thank you for using Student Score Analyzer!")
            break
        else:
            print("Invalid choice! Please try again.")


main()
```
